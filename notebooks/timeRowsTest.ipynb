##Workaround for font limit and printing the number of sheets

```python
from openpyxl.styles.fonts import Font
Font.family.max = 100

from openpyxl import load_workbook

wb = load_workbook("../data/raw/Nutrition141025NEW.xlsx",
                   read_only=True, data_only=True)
print(len(wb.sheetnames))
```

## Tells us which row the time column is in, and which columns

```python
DAYS = {"Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"}

ws = wb["06JUL26"]

for row_number, cell_values in enumerate(ws.iter_rows(values_only=True), start=1): #For loop begins,
    time_columns = []
    for column_number, value in enumerate(cell_values):
        if str(value).strip() == "Time":
            time_columns.append(column_number)
    if time_columns:
        print("row", row_number, "-> Time at columns", time_columns)
```
